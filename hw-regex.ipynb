```python
## import libaries
import pandas as pd
import regex as re
import glob
from mangoCR import pdf2image_ocr_text
```

```python
## Defining regex patterns
pattern_case_num = r'#:\s\d+\w'

pattern_request_date = r':\W\w+\W\d+,\W\d+$'

pattern_decision_date = r'\d+\/\d+\/\d+'
```

```python
## glob some pdfs together- using this code because normally you'd have to many docs to go through, not just one.
target_pdfs = glob.glob("Redacted_*pdf")
target_pdfs
```

```python
## convert pdf to text (not markdown)
# text = pdf2image_ocr_text(target_pdfs[-1])
# text
text = pdf2image_ocr_text(target_pdfs[-1])
text
```

```python
case_numbers = re.findall(r'#:\s\d+\w', text)
case_numbers
```

```python
request_dates = re.findall(r':\W\w+\W\d+,\W\d+$', text, re.MULTILINE)
request_dates
```

```python
decision_dates = re.findall(r'\d+\/\d+\/\d+', text)
decision_dates
```
