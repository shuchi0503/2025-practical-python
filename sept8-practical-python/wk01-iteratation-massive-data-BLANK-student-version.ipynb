```python
import pandas as pd
```

```python
#check out the top ten
df_head = pd.read_csv("EMS_Incident_Dispatch_Data_20240407.csv", nrows = 12)
df_head
```

```python
## create a list using chat-gpt of how many of these are covid related

related_to_covid = [
    "ABDFC", "ABDPFC", "ABDPFT",
    "ALTMFC", "ALTMFT",
    "ANAPFC", "ANAPFT",
    "ARREFC", "ARREFT",
    "ASTHFC", "ASTHFT",
    "CARDFC", "CARDFT",
    "CHOKFC", "CHOKFT",
    "CVACFC", "CVACFT", "CVAFC", "CVAFT",
    "DIFFFC", "DIFFFT",
    "DRUGFC",
    "FEVER",
    "INBLFC", "INBLFT",
    "MEDRFC",
    "PEDFC", "PEDFT",
    "RESPFC", "RESPFT",
    "SEIZFC", "SEIZFT",
    "SICKFC", "SICKFT",
    "STATFC", "STATFT",
    "UNCFC", "UNCFT"
]
```

```python
#length of possible covid list - lesson- do it manually
len(related_to_covid)
```

```python
# chunking- we have to save each one as a file

row_size = 500_000

df_filtered = []

for i, a_chunk in enumerate(pd.read_csv("EMS_Incident_Dispatch_Data_20240407.csv", 
                                        chunksize = row_size, low_memory = False), start = 1):
    print(f"Analysing chunk no. {i}")
    df_filtered.append(a_chunk.query("FINAL_CALL_TYPE in @related_to_covid"))

print("Your filtered df is read to analyze")
    
```
